## 🔹 Paso 2: Preparación de los datos

Durante la preparación del dataset se crearon los campos calculados necesarios para medir la rentabilidad del negocio y construir los indicadores del dashboard.

**Columna calculada: Utilidad**

Se creó el campo calculado Utilidad, definido como la diferencia entre los Ingresos y el Costo de cada transacción.

Fórmula utilizada en Tableau:

SUM([Ingresos]) - SUM([Costo]) 

Esta métrica representa la ganancia generada por cada venta y fue utilizada como uno de los principales indicadores de desempeño comercial en ambos dashboards.

Indicador calculado: Margen %

También se creó el indicador Margen %, calculado como la relación entre la utilidad y los ingresos.

Fórmula utilizada en Tableau:

SUM([Utilidad]) / SUM([Ingresos])

El resultado se formateó como porcentaje con un decimal, permitiendo comparar la rentabilidad entre países, categorías y segmentos de clientes independientemente del volumen de ventas.
